```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Send an HTTP request to the target public web page
url = "http://books.toscrape.com/"
response = requests.get(url)
response.encoding = 'utf-8'

# 2. Parse the HTML structure
soup = BeautifulSoup(response.text, 'html.parser')

# 3. Extract relevant data elements (Book titles and prices)
books = []
for item in soup.find_all('article', class_='product_pod'):
    title = item.find('h3').find('a')['title']
    price = item.find('p', class_='price_color').text
    books.append({'Book Title': title, 'Price': price})

# 4. Create a custom dataset DataFrame
df = pd.DataFrame(books)
print(df.head())

# Save dataset to CSV for your analysis needs
df.to_csv("codealpha_scraped_books.csv", index=False)
```
